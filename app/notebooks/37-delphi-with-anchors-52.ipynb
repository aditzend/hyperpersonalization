```python
import os
from dotenv import load_dotenv, find_dotenv

# 1. find_dotenv(): Busca el archivo .env automáticamente subiendo directorios hasta encontrarlo.
# 2. override=True: IMPORTANTE. Esto fuerza a actualizar la variable si ya existía en memoria.
exito = load_dotenv(find_dotenv(), override=True)

if exito:
    print("✅ Variables de entorno recargadas correctamente.")
else:
    print("❌ No se encontró el archivo .env.")

# print(os.getenv("OPENAI_API_KEY"))
```

```python
from openai import OpenAI
from datetime import datetime

# 1. Configuración del cliente
# Se recomienda usar una variable de entorno, pero puedes pegar tu key aquí si es para pruebas.
# client = OpenAI(api_key="TU_CLAVE_API_AQUI")

def ejecutar_delphi_simulado(system_prompt, user_prompt, model, temperature, top_p, seed):
    """
    Ejecuta una llamada a la API de OpenAI para simular el método Delphi.
    Parámetros:
        system_prompt (str): Mensaje para el rol 'system'
        user_prompt (str): Mensaje para el rol 'user'
        model (str): Nombre del modelo (por ejemplo, "gpt-4o")
        temperature (float): Valor de temperatura para generación
        top_p (float): Valor de top_p para generación
        seed (int): Semilla para reproducibilidad (best-effort)
    Retorna:
        tuple: (respuesta_texto, fingerprint, response_dict)
    """
    client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
    try:
        response = client.chat.completions.create(
            model=model, # e.g., "gpt-4o"
            temperature=temperature,
            top_p=top_p,
            seed=seed,
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ]
        )
        respuesta_texto = response.choices[0].message.content




        fingerprint = response.system_fingerprint
        # if fingerprint is None:
        #     print("\n⚠️ El API devolvió system_fingerprint como None.")
        #     print("Esto significa que no hay información de configuración del backend disponible para esta petición.")
        # else:
        #     print(f"\n✅ System Fingerprint recibido: {fingerprint}")

        return respuesta_texto, fingerprint, response.model_dump()
    except Exception as e:
        print(f"Ocurrió un error: {e}")
        return None, None, None
```
